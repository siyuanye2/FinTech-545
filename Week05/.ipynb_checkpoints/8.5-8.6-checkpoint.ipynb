#### 8.5
Calculate ES from T Distribution, using test7_2.csv and testout_8.5.csv

```python
import numpy as np
import pandas as pd
from scipy.stats import t
```

```python
data = np.loadtxt('../testfiles/data/test7_2.csv', skiprows=1)
data
```

```python
est_nu, est_mu, est_sigma = t.fit(data)

alpha = 0.05
t_stat = t.ppf(alpha, df=est_nu)
phi_t = t.pdf(t_stat, df=est_nu)
```

```python
es_abs = -est_mu + est_sigma * ((est_nu + t_stat**2) / (est_nu - 1)) * (phi_t / alpha)
es_diff = est_sigma * ((est_nu + t_stat**2) / (est_nu - 1)) * (phi_t / alpha)

print(f"ES Absolute,ES Diff from Mean")
print(f"{es_abs},{es_diff}")
```

#### 8.6
Calculate ES from Simulation -- compare to 8.5 values, using test7_2.csv and testout_8.6.csv

```python
np.random.seed(42)
n_sims = 100_000

# draw random num from T dist, 100,000 x 5
t_sim = t.rvs(est_nu, est_mu, est_sigma, size=n_sims)
```

```python
var_abs = np.quantile(t_sim, alpha)
es_abs = abs(t_sim[t_sim <= var_abs].mean())

es_diff = est_mu - t_sim[t_sim <= var_abs].mean()

print(f"ES Absolute,ES Diff from Mean")
print(f"{es_abs},{es_diff}")
```
