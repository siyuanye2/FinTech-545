Calculate ES From Normal Distribution, using test7_1.csv and testout_8.4.csv

```python
import numpy as np
from scipy.stats import norm
```

```python
data = np.loadtxt('../testfiles/data/test7_1.csv', skiprows=1)
data
```

```python
est_mu = np.mean(data)
est_sigma = np.std(data, ddof=1)

alpha = 0.05
z_score = norm.ppf(alpha)
phi_z = norm.pdf(z_score)
```

```python
es_abs = -est_mu + est_sigma * phi_z / alpha
es_diff = est_sigma * phi_z / alpha

print(f"ES Absolute,ES Diff from Mean")
print(f"{es_abs},{es_diff}")
```
