```python
import pandas as pd
import numpy as np
```

#### 6.1
calculate arithmetic returns using test6.csv and testout6.1.csv

```python
# read input data
data = pd.read_csv('../testfiles/data/test6.csv')

data["Date"] = pd.to_datetime(data["Date"])
data = data.sort_values("Date")
data = data.set_index("Date")

data
```

```python
# arithmetic return
returns = data.pct_change()
returns = returns.dropna()
returns
```
