```python
import pandas as pd
```

#### 1.1
Calculate covariance skipping missing rows, using test1.csv and testout_1.1.csv

```python
# read input data
data = pd.read_csv('../testfiles/data/test1.csv')
data_clean = data.dropna()
```

```python
print(data.shape)   # (10, 5)
print(data_clean.shape)   # (4, 5)
```

```python
# cov
cov_matrix = data_clean.cov()
print(cov_matrix)
```

#### 1.2
Calculate correlation skipping missing data, using test1.csv and testout_1.2.csv

```python
# corr
corr_matrix = data_clean.corr()
print(corr_matrix)
```

#### 1.3
Calculate pairwise covariance, using test1.csv and testout_1.3.csv

```python
pairwise_cov = data.cov()
print(pairwise_cov)
```

#### 1.4
Calculate pairwise correlation, , using test1.csv and testout_1.4.csv

```python
pairwise_corr = data.corr()
print(pairwise_corr)
```
