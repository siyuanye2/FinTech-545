Calculate VaR from Normal Distribution, using test7_1.csv and testout8_1.csv)

```python
import numpy as np
from scipy.stats import norm
```

```python
data = np.loadtxt('../testfiles/data/test7_1.csv', skiprows=1)
data
```

```python
est_mu = np.mean(data)
est_sigma = np.std(data, ddof=1)
```

```python
alpha = 0.05
z_score = norm.ppf(alpha)
```

```python
var_percentile = est_mu + z_score * est_sigma

var_abs = abs(var_percentile)
var_diff = est_mu - var_percentile

print(f"VaR Absolute,VaR Diff from Mean")
print(f"{var_abs},{var_diff}")
```
