#### 8.2
Calculate VaR from T Distribution, using test7_2.csv and testout_8.2.csv

```python
import numpy as np
import pandas as pd
from scipy.stats import t
```

```python
data = np.loadtxt('../testfiles/data/test7_2.csv', skiprows=1)
data
```

```python
est_nu, est_mu, est_sigma = t.fit(data)
```

```python
alpha = 0.05
t_stat = t.ppf(alpha, df=est_nu)
```

```python
var_percentile = est_mu + t_stat * est_sigma

var_abs = abs(var_percentile)
var_diff = est_mu - var_percentile

print(f"VaR Absolute,VaR Diff from Mean")
print(f"{var_abs},{var_diff}")
```

#### 8.3
Calculate VaR from T Simulation -- compare to 8.2 values using test7_2.csv and testout_8.3.csv

```python
np.random.seed(42)
n_sims = 100_000
```

```python
# draw random num from T dist, 100,000 x 5
t_sim = t.rvs(est_nu, est_mu, est_sigma, size=n_sims)
```

```python
var_abs = abs(np.quantile(t_sim, alpha))
var_diff = est_mu - np.quantile(t_sim, alpha)

print(f"VaR Absolute,VaR Diff from Mean")
print(f"{var_abs},{var_diff}")
```
